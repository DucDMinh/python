```python
class HinhChuNhat:
    def __init__(self, width, height):
        self.width = width
        self.height = height
    def Perimeter(self):
        return (self.width + self.height) * 2
    def Area(self):
        return self.width * self.height
    def display(self):
        print(f"Chieu dai: {self.height}, Chieu rong: {self.width}, Chu vi: {self.Perimeter()}, Dien tich: {self.Area()}")
hcn = HinhChuNhat(5, 6)
hcn.display()
```
