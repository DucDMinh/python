```python
import random

a = int(input("Nhap so nguyen a: "))
b = int(input("Nhap so nguyen b: "))

print("Kết quả:", end='\n')
print('Tổng: ', a + b, end='\n')
print("Hiệu: ", a - b, end='\n')
print('Tích: ', a * b, end='\n')
print("Thương: ", float(a / b), end='\n')
```

```python
import math

x1 = int(input("Nhập tọa độ x1 của điểm A: "))
y1 = int(input("Nhập tọa độ y1 của điểm A: "))
x2 = int(input("Nhập tọa độ x2 của điểm B: "))
y2 = int(input("Nhập tọa độ y2 của điểm B: "))

e = math.sqrt((x1 - x2)**2 + (y1 - y2)**2)

print("Khoảng cách Euclidean giữa A và B: ", e)
```

```python
n = int(input("Nhập một số nguyên có ít hơn 5 chữ số: "))
if 0 <= n < 10000:
    ngan = n // 1000
    tram = (n % 1000) // 100
    chuc = (n % 100) // 10
    don_vi = n % 10

    result = ""
    if ngan > 0:
        result += f"{ngan} ngàn "
    if tram > 0 or ngan > 0:
        result += f"{tram} trăm "
    if chuc > 0 or tram > 0 or ngan > 0:
        result += f"{chuc} chục "
    result += f"{don_vi} đơn vị"

    print(f"Số {n} đọc là: {result.strip()}")
else:
    print("Số nhập vào không hợp lệ.")
```
