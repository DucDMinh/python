```python
def tim_hoan_vi(a, b, c):
    # Đưa 3 số vào một danh sách để dễ lấy qua chỉ mục (index)
    ds = [a, b, c]

    print(f"Các hoán vị của 3 số {a}, {b}, {c} là:")

    # Dùng 3 vòng lặp duyệt qua các vị trí (0, 1, 2)
    for i in range(3):
        for j in range(3):
            for k in range(3):
                # Chỉ in ra khi 3 vị trí (index) khác nhau hoàn toàn
                if i != j and i != k and j != k:
                    print((ds[i], ds[j], ds[k]))

# Chạy thử nghiệm
tim_hoan_vi(1, 2, 3)
```

```python
import string
def input_string():
    s = str(input("Enter a sentence: "))
    lower_s = s.lower()
    clean_sentence = ""
    for char in lower_s:
        if char not in string.punctuation:
            clean_sentence += char
    print("Lower case sentence:", lower_s)
    print("Clean case sentence:", clean_sentence)

    word_list = clean_sentence.split()
    count_dic = {}
    for word in word_list:
        if word not in count_dic:
            count_dic[word] = 1
        else :
            count_dic[word] += 1
    count_dic = sorted(count_dic.items(), key=lambda x: x[1], reverse=True)
    print("----Result----")
    for word, count in count_dic:
        print(f"{word}: {count}")
input_string()
```
